```python
import numpy as np
import matplotlib.pyplot as plt 
import statistics 

#Calculating Uncertainty on Time(Down) Measurements
t_down = [15.06, 15.26, 16.91, 7.9, 20.14, 21.44, 38.42, 22.86, 6.26, 21.68, 17.72, 21.18,
     20.52, 9.24, 16.82, 11.6, 12.72, 10.16, 23.26]
n = 19
t_down_avg = statistics.mean(t_down)
t_down_sd = statistics.stdev(t_down)
def delta_t_down(t_down_sd, n):
    return t_down_sd/np.sqrt(n)

print(delta_t_down(t_down_sd, n))
```

```python
#Calculating Uncertainty on Velocity(Down) Measurements

v_down = [6.64e-5, 6.55e-5, 5.91e-5, 1.26e-4, 4.97e-5,4.66e-5,
     2.60e-5, 4.37e-5, 1.60e-4, 4.61e-5, 5.64e-5, 4.72e-5, 4.87e-5,
     1.08e-4, 5.95e-5, 8.62e-5, 7.86e-5, 9.84e-5, 4.30e-5]


def delta_v_down(v_down, delta_t_down, t_down):
    deltas = []
    for i in range(len(v_down)):
        deltas.append(v_down[i] * (delta_t_down / t_down[i]))
    return deltas

print(delta_v_down(v_down, delta_t_down(t_down_sd, n), t_down))
```

```python
#Calculating Uncertainty on Time(Up) Measurements

t_up = [2.85, 0.65, 0.65, 6.44, 0.92, 3.4, 10.58, 0.62, 1.86, 
        0.84, 2.06, 2.46, 5.82, 3.7, 2.98, 5.88, 2.18, 3.18, 2.56]
n = 19
t_up_avg = statistics.mean(t_up)
t_up_sd = statistics.stdev(t_up)

def delta_t_up(t_up_sd, n):
    return t_up_sd/np.sqrt(n)

print(delta_t_up(t_up_sd, n))
```

```python
#Calculating Uncertainty on Velocity(Up) Measurements
v_up = [3.51e-4, 1.54e-3, 1.54e-3, 1.55e-4, 1.09e-3, 2.94e-4, 9.45e-5, 1.61e-3,
        5.38e-4, 1.19e-3, 4.85e-4, 4.07e-4, 1.72e-4, 2.7e-4, 3.36e-4, 1.7e-4, 4.59e-4,
        3.14e-4, 3.91e-4]

def delta_v_up(v_up, delta_t_up, t_up):
    deltas = []
    for i in range(len(v_up)):
        deltas.append(v_up[i] * (delta_t_up / t_up[i]))
    return deltas

print(delta_v_up(v_up, delta_t_up(t_up_sd, n), t_up))
```

```python
#Graph Comparing Up and Down Velocities with error bars

import matplotlib.pyplot as plt

# Plotting down data
plt.errorbar(range(len(v_down)), v_down, yerr=delta_v_down(v_down, delta_t_down(t_down_sd, n), t_down),
             fmt='o', capsize=10, label='Down')

# Plotting up data
plt.errorbar(range(len(v_up)), v_up, yerr=delta_v_up(v_up, delta_t_up(t_up_sd, n), t_up),
             fmt='o', capsize=3, label='Up')

plt.xlabel("Trial")
plt.ylabel("Velocity (m/s)")
plt.title("Up vs Down Velocities with Uncertainties")
plt.legend()
plt.show()

#Plotting Down Velocity by itself.
plt.figure()
plt.errorbar(range(1, 20), v_down, yerr=delta_v_down(v_down, delta_t_down(t_down_sd, n), t_down),
             fmt='o', capsize=5, label='Down')
plt.xlabel("Trial")
plt.ylabel("Velocity (m/s)")
plt.title("Down Velocity with Uncertainties")
plt.legend()
plt.show()
```
