```python
#Double slit Exp(1)
#Plotting mA vs theta to show angles at which maxima and minima occur
#For spacer = 6cm
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0,86,5) #degrees
y = np.array([0.6, 0.6, 0.1, 0.22, 0.4, 0.4, 0.3,  #current 
              0.08, 0.06, 0.2, 0.27, 0.25, 0.15, 
              0.1, 0.03, 0.01, 0,0])

plt.scatter(x, y)
plt.plot(x,y)
plt.xlabel("Angles in Degrees")
plt.ylabel("Current in mA")
plt.title("Current vs Theta")

#theoretical values
maxima_theoretical = [0, 22.33, 49.46]
minima_theoretical = [10.93, 34.75, 71.81]

for x in maxima_theoretical:
    plt.axvline(x, color='red', linestyle=':',
                label='maxima' if x == maxima_theoretical[0] else '_nolegend_')

for x in minima_theoretical:
    plt.axvline(x, color='black', linestyle=':',
                label='minima' if x == minima_theoretical[0] else '_nolegend_')

plt.legend()
```

```python
#Bragg Diffraction 
#Graphing relative frequency vs grazing angle.

theta = np.arange(0,61,3)
intensity = np.array([0.6, 0.6, 0.58, 0.51, 
                      0.42, 0.36, 0.32, 0.36,
                      0.42, 0.46, 0.34, 0.08, 
                      0.06, 0.22, 0.3, 0.22, 
                      0.06, 0.2, 0.34,
                      0.4, 0.26])
maxima = np.array([27, 42, 57])

plt.plot(theta, intensity)
for x in maxima:
    plt.axvline(x, color='orange', linestyle=':',
                label='maxima' if x == maxima[0] else '_nolegend_')

plt.legend()
plt.scatter(theta, intensity)
plt.title("Intensity of Diffracted Signal vs Grazing Angle")
plt.xlabel("Grazing Angle")
plt.ylabel("Intensity")
```
