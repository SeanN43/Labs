```python
#Sample Data 
#Droplet 1A

#Average Velocity for droplet 1A (Down)
import numpy as np
import matplotlib.pyplot as plt 
import statistics 

t_Down = [18.24, 18.56, 19.24, 18.05, 17.23, 15.35, 16.7, 17.99, 15.35, 17.25, 18.38]
d_Down = [5e-4, 5e-4, 5e-4, 5e-4, 5e-4, 5e-4, 5e-4, 5e-4, 5e-4, 5e-4, 5e-4]

def v_Down(t_Down, d_Down):
    return [d_Down[i] / t_Down[i] for i in range(len(t_Down))]

print(v_Down(t_Down, d_Down))
print(statistics.mean(v_Down(t_Down, d_Down)))
```

```python
#Uncertainties for Time(down)
n = 11
t_Down_avg = statistics.mean(t_Down)
print(statistics.mean(t_Down))
t_Down_sd = statistics.stdev(t_Down)

def delta_t_Down(t_Down_sd, n):
    return t_Down_sd/np.sqrt(n)

print(delta_t_Down(t_Down_sd, n))

#Uncertainty for Velocity(down)
v_Down = [2.741228070175439e-05, 2.6939655172413796e-05, 2.598752598752599e-05, 2.7700831024930747e-05,
          2.901915264074289e-05, 3.2573289902280134e-05, 2.9940119760479045e-05, 2.7793218454697057e-05,
          3.2573289902280134e-05, 2.8985507246376814e-05, 2.720348204570185e-05]

def delta_v_Down(v_Down, delta_t_Down, t_Down):
    deltas = []
    for i in range(len(v_Down)):
        deltas.append(v_Down[i] * (delta_t_Down / t_Down[i]))
    return deltas

print(delta_v_Down(v_Down, delta_t_Down(t_Down_sd, n), t_Down))
```

```python
#Average Velocity (Up)

t_Up = [3.89, 11, 11.59, 11.17]
d_Up = [5e-4, 1.5e-3, 1.5e-3, 1.5e-3]

def v_Up(t_Up, d_Up):
    return [d_Up[i] / t_Up[i] for i in range(len(t_Up))]

print(v_Up(t_Up, d_Up))
print(statistics.mean(v_Up(t_Up, d_Up)))
```

```python
#Uncertainties for Time(Up)
n2 = 4
t_Up_avg = statistics.mean(t_Up)
print(statistics.mean(t_Up))
t_Up_sd = statistics.stdev(t_Up)

def delta_t_Up(t_Up_sd, n):
    return t_Up_sd/np.sqrt(n2)

print(delta_t_Up(t_Up_sd, n2))
```

```python
#Uncertainty for Velocity(Up)
v_Up = [1.29e-4, 1.36e-4, 1.29e-4, 1.34e-4]

def delta_v_Up(v_Up, delta_t_Up, t_Up):
    deltas = []
    for i in range(len(v_Up)):
        deltas.append(v_Up[i] * (delta_t_Up / t_Up[i]))
    return deltas

print(delta_v_Up(v_Up, delta_t_Up(t_Up_sd, n), t_Up))
```
